```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
#print(PROJECT_ROOT)   
DATA_DIR = PROJECT_ROOT / "data" / "raw"

print(DATA_DIR)
```

```python
# 현재 작업 폴더에서 위로 올라가며 data 폴더를 찾는다.
current = Path.cwd().resolve()
print("시작 위치:", current)

project_root = None
while True:
    if (current / "data").is_dir():
        project_root = current      # data 폴더가 있으면 여기가 프로젝트 루트
        break
    if current == current.parent:   # 더 이상 위로 못 올라가면(디스크 최상위) 중단
        break
    current = current.parent        # 부모 폴더로 한 단계 위로

if project_root is None:
    raise FileNotFoundError("프로젝트 루트를 찾을 수 없습니다.")

print("프로젝트 루트:", project_root)
```

```python
# 위 로직을 함수로 정리한 것 (course_utils/paths.py 에 그대로 옮길 코드)
from pathlib import Path


def get_project_root() -> Path:
    current = Path(__file__).resolve().parent if "__file__" in dir() else Path.cwd().resolve()
    while current != current.parent:
        if (current / "data").is_dir():
            return current
        current = current.parent
    raise FileNotFoundError("프로젝트 루트를 찾을 수 없습니다.")


def get_data_dir() -> Path:
    return get_project_root() / "data"


print("프로젝트 루트:", get_project_root())
print("데이터 폴더:", get_data_dir())
```
