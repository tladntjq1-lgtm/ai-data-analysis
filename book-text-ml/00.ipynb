## 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기
원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

이번 실습에서는 다음처럼 간단하게 정리할 수 있습니다.


```python
# 컬럼 이름 일부를 딕셔너리로 매핑해서 변경
df_books = df_books.rename(columns={
    '판매상품 ID': '판매상품ID',
    '인물': '저자',
    '발행(출시)일자': '발행일'
})

df_books.columns  # 변경된 컬럼 이름 확인
```

## 컬럼 이름을 사용하기 쉽게 정리하기

`rename()`을 사용해 `df_books`의 컬럼 이름 중 일부를 더 짧고 간단한 이름으로 변경했다.

- 판매상품 ID -> 판매상품ID

- 발행(출시)일자 -> 발행일

변경 후 `df_books.columns`로 확인한 결과, 지정한 3개 컬럼만 이름이 바뀌고 나머지 컬럼(순위, 상품명, 판매가, 출판사, 분야)은 그대로 유지된 것을 확인했다. 공백이나 괄호가 없는 이름으로 정리되어 앞으로 코드에서 컬럼을 다룰 때 더 편해질 것으로 보인다.

## 실습 5. 결측치 확인하기


결측치는 값이 비어 있는 데이터를 의미합니다.

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.


```python
# 컬럼별 결측치 개수 확인
df_books.isna().sum()
```

```python
# 결측치 개수 세기
missing = df_books.isna().sum()

# 결측치가 1개 이상인 컬럼만 보기
missing[missing > 0]
```

## 결측치 확인하기

`df_books.isna().sum()`으로 컬럼별 결측치 개수를 확인했다. 순위, 판매상품ID, 상품명, 판매가는 결측치가 없었고, 저자·출판사·발행일은 각각 3개, 분야는 62개의 결측치가 있었다.

결측치가 1개 이상인 컬럼만 `missing[missing > 0]`으로 걸러서 확인한 결과 저자, 출판사, 발행일, 분야 4개 컬럼에만 결측치가 존재함을 다시 한 번 확인했다. 특히 분야 컬럼은 결측치 비율이 눈에 띄게 높아, 이후 분석 시 별도의 처리 방법을 고민해야 할 것으로 보인다.

## 실습 6. 판매가를 숫자로 바꾸기

 

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.


```python
# 1. 쉼표(,)를 빈 문자열로 바꿔서 제거
df_books['판매가'] = df_books['판매가'].str.replace(',', '')

# 2. 문자열을 숫자형(정수)으로 변환
df_books['판매가'] = df_books['판매가'].astype(int)

# 3. 변환된 dtype 확인
df_books['판매가'].dtype
```

## 판매가를 숫자로 바꾸기

`판매가` 컬럼이 `'16,200'`처럼 쉼표가 포함된 문자열로 저장되어 있어 계산에 바로 사용할 수 없었다.

`str.replace(',', '')`로 쉼표를 제거한 뒤 `astype(int)`로 정수형으로 변환했고, 변환 후 dtype을 확인한 결과 `int64`로 정상 변환된 것을 확인했다.

이제 판매가 컬럼으로 합계, 평균, 정렬 등 숫자 기반 분석이 가능해졌다.

## 실습 7. 발행일을 날짜형으로 바꾸기

 

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# 1. float 타입의 발행일을 정수형(Int64, 결측치 허용)으로 바꾼 뒤 문자열로 변환
df_books['발행일'] = df_books['발행일'].astype('Int64').astype(str)

# 2. YYYYMMDD 형식의 문자열을 날짜형으로 변환 (형식이 안 맞으면 NaT 처리)
df_books['발행일'] = pd.to_datetime(df_books['발행일'], format='%Y%m%d', errors='coerce')

# 3. 변환된 dtype과 앞의 5개 값 확인
print(df_books['발행일'].dtype)
df_books['발행일'].head()
```

## 발행일을 날짜형으로 바꾸기

`발행일` 컬럼이 `20260701.0`처럼 실수형(float)으로 저장되어 있어 날짜로 다루기 어려웠다.

먼저 `Int64`로 변환해 결측치를 유지한 채 정수로 바꾸고, 다시 문자열로 변환한 뒤 `pd.to_datetime()`에 `format='%Y%m%d'`를 지정해 날짜형으로 변환했다. `errors='coerce'` 옵션 덕분에 형식이 맞지 않거나 결측치인 값은 오류 없이 `NaT`로 처리되었다.

변환 결과 dtype이 `datetime64[us]`로 바뀐 것을 확인했고, 앞의 5개 값을 보면 2026-07-01, 2026-08-31, 2002-01-20, 2026-09-16, 2026-09-16처럼 실제 날짜 형태로 잘 표시되었다.


##  실습 8. 문자열 앞뒤 공백 정리하기
상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.


```python
# 공백을 제거할 컬럼 목록
cols_to_strip = ['상품명', '저자', '출판사', '분야']

# 반복문으로 각 컬럼에 str.strip() 적용
for col in cols_to_strip:
    df_books[col] = df_books[col].str.strip()

# 처리 후 값 확인 (컬럼 4개를 한 번에 5행씩)
df_books[cols_to_strip].head()
```

## 문자열 앞뒤 공백 제거하기

`상품명`, `저자`, `출판사`, `분야` 컬럼에 눈에 보이지 않는 앞뒤 공백이 섞여 있으면, 실제로는 같은 값인데도 서로 다른 문자열로 인식되어 그룹핑이나 집계 시 오류가 생길 수 있다.

반복문으로 4개 컬럼에 각각 `str.strip()`을 적용해 앞뒤 공백을 제거했다. 처리 후 값을 확인한 결과 데이터 자체는 그대로 유지되면서 불필요한 공백만 정리된 것을 확인했다.

이 작업으로 이후 분야별 집계나 저자·출판사 기준 그룹화 작업에서 공백 때문에 생기는 오류를 예방할 수 있다.

## 실습 9. 중복 데이터 확인하기

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

 
먼저 무엇을 기준으로 중복인지 생각해야 합니다.


예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.



하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

 
따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

 
우선 상품을 구분하는 판매상품ID를 확인합니다.


```python
# 1. 전체 행이 완전히 같은 중복 행 개수
df_books.duplicated().sum()
```

```python
# 2. 판매상품ID가 중복된 행 개수
df_books.duplicated(subset='판매상품ID').sum()
```

```python
# 3. 상품명이 중복된 행 개수
df_books.duplicated(subset='상품명').sum()
```

## 중복 데이터 확인하기

`duplicated()`를 이용해 세 가지 기준으로 중복 여부를 확인했다.

- 전체 행이 완전히 같은 중복: 0건
- 판매상품ID 기준 중복: 0건
- 상품명 기준 중복: 37건

판매상품ID는 상품을 구분하는 고유 식별자이므로 중복이 없는 것이 정상이며, 실제로도 중복이 확인되지 않았다. 반면 상품명 기준으로는 37건의 중복이 발견되었는데, 이는 같은 책이 다른 판매상품ID(예: 다른 판본, 세트 구성 등)로 여러 번 등록되었을 가능성을 의미한다. 상품을 구분할 때는 상품명이 아니라 판매상품ID를 기준으로 삼아야 한다는 것을 알 수 있다.

## 14. 간단한 확인 문제

**질문 1. 왜 모든 원본 컬럼을 사용하지 않고 필요한 컬럼만 선택했나요?**

원본 데이터에는 정가, 할인율, 적립율, 적립예정포인트처럼 이번 분석 목적과 크게 관련 없는 컬럼도 섞여 있었다. 불필요한 컬럼까지 모두 들고 있으면 데이터를 다루기 복잡해지고, 실수로 관련 없는 값을 분석에 끼워 넣을 위험도 커진다. 필요한 컬럼만 골라두면 이후 코드가 더 단순해지고 실수도 줄어든다.

**질문 2. 판매가가 문자열이라면 어떤 문제가 발생할 수 있나요?**

문자열 상태로는 합계, 평균, 정렬 같은 숫자 계산이 불가능하거나 엉뚱한 결과가 나올 수 있다. 예를 들어 `'16,200'`처럼 쉼표가 들어 있으면 숫자로 인식되지 않아 계산 시 오류가 나거나, 문자열 기준으로 정렬되어 크기 순서가 실제 숫자 순서와 다르게 나올 수 있다.

**질문 3. 결측치를 확인한 뒤 바로 모든 행을 삭제하면 안 되는 이유는 무엇인가요?**

결측치가 있는 컬럼과 개수는 컬럼마다 다르다(예: 분야는 62개, 저자는 3개). 결측치가 있는 행을 무조건 다 지우면 분야 결측 62건 때문에 저자·출판사 정보는 멀쩡한 데이터까지 함께 사라져서 데이터 손실이 커진다. 어떤 컬럼의 결측치가 분석에 실제로 중요한지 먼저 판단한 뒤, 필요한 경우에만 제거하거나 다른 방식으로 처리해야 한다.

**질문 4. 같은 상품명이 두 번 등장했다고 해서 바로 중복 데이터라고 판단하기 어려운 이유는 무엇인가요?**

같은 책이라도 판형, 세트 구성, 개정판 등으로 인해 상품명은 같지만 실제로는 서로 다른 상품(판매상품ID가 다름)인 경우가 있을 수 있다. 실제로 상품명 기준 중복은 37건이었지만 판매상품ID 기준 중복은 0건이었던 것처럼, 상품명만으로 중복을 판단하면 서로 다른 상품을 같은 것으로 잘못 처리할 수 있다.

**질문 5. 인공지능이 작성한 코드가 오류 없이 실행되었다면 그것만으로 분석이 끝난 것일까요?**

아니다. 코드가 오류 없이 실행되는 것과 결과가 의도한 대로 맞게 나온 것은 다른 문제다. 변환된 dtype이 예상한 자료형인지, 결측치나 중복 개수가 합리적인지 직접 눈으로 확인하고 그 의미를 이해해야 분석이 제대로 끝났다고 할 수 있다.
